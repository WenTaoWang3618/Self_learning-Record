```python
'''
面积计算 area
    输入: r, constant
        r -> 半径
        constant -> 判定形状(正方形,圆形,六边形)
    
    Process:
        area_square: 正方形面积
        area_cir: 圆形面积
        area_hexagon: 六边形面积
    
    输出: area
        total -> 累加的和
'''
def area(r, constant):
    assert r>0, 'A length must be positive'
    return r*r*constant

def area_square(r):
    return area(r,1)

def area_cir(r):
    return area(r,pi)

def area_hexagon(r):
    return area(r,3*sqrt(3)/2)
```

```python
'''
不同法则的累加summation
    输入: k, term
        k -> 累加个数
        term -> 累加法则
    
    Process:
        summary: 加和
        cube: 立方加和
        pai: Pi计算加和
    
    输出: total
        total -> 累加的和
'''
def summation(k,term):
    total = 0
    for i in range(k+1):
        print(total, i)
        total += term(i)
    return total

def summary(n):
    return n

def cube(n):
    return n**3

def pai(n):
    return 8/((4*k-3)*(4*k-1))

summation(5,cube)
```

```python
# 内置函数
def make_adder(n):
    def adder(k):
        return n + k
    return adder

add_three = make_adder(3) # add_three变成一个每次加3的函数

add_three(4) # 4 + 3

make_adder(5)(4) # 5 + 4
```

```python
'''
Python 装饰器（Decorator）
    Python 装饰器（Decorator） 本质上是一个接收函数作为输入、并返回一个新函数的“包装器”。
    它的核心作用是：在不修改原函数定义与调用方式的前提下，给函数动态增加额外的功能
    赋予一个h
'''

import time
from functools import wraps

def timer(func):
    @wraps(func)  # 关键点：保留原函数的名称、文档等元数据
    def wrapper(*args, **kwargs):
        # 1. 在原函数执行前做点事
        start_time = time.time()
        
        # 2. 执行真正的原函数（*args, **kwargs 保证能接收任意参数）
        result = func(*args, **kwargs)
        
        # 3. 在原函数执行后做点事
        end_time = time.time()
        print(f"⏱️ 函数 [{func.__name__}] 耗时: {end_time - start_time:.4f} 秒")
        
        # 4. 返回原函数的计算结果
        return result
    
    return wrapper  # 返回包装好的新函数
```

```python
@timer
def compute_sum(n):
    """计算 0 到 n-1 的累加和"""
    return sum(range(n))

# 调用方式与普通函数完全一致
total = compute_sum(10000000)
# 自动输出：⏱️ 函数 [compute_sum] 耗时: 0.1823 秒
```
